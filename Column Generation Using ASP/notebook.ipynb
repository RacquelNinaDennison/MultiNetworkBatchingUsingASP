night ← (x < 7).
night ← (x ≥ 22).
```

The constraint atom `(x < 7)` appears in a body, but **never in a head**. 

### Standard Completion Says:

For every atom `a`, we need a nogood that says:
> "If `a` is true, then at least one of its supporting bodies must be true"

$$\delta_P(a) = \{Ta, F\beta_1, ..., F\beta_k\}$$

where β₁, ..., βₖ are all bodies that have `a` as their head.

### For `(x < 7)`:

- How many rules have `(x < 7)` as the head? **Zero!**
- So: `body_P((x < 7)) = {}` (empty set)
- The nogood becomes: `δ_P((x < 7)) = {T(x < 7)}`

### What Does This Nogood Mean?
```
{T(x < 7)}
```

This is a **unit nogood** — it has only one literal. It says:

> "The assignment T(x < 7) is forbidden"

So immediately, `(x < 7)` is forced to be **false**!

### This Is Wrong!

Whether `(x < 7)` is true should depend on the value of `x`:
- If `x = 3`, then `(x < 7)` should be **true**
- If `x = 10`, then `(x < 7)` should be **false**

The program rules shouldn't force it either way.

---

## The Solution: External Atoms

We designate constraint atoms as **external** — their truth comes from **outside** the logic program.

### Two Types of Atoms

| Type | Symbol | Truth Determined By |
|------|--------|---------------------|
| **Regular atoms** | 𝒜 | Rules in the program |
| **External atoms** | 𝒞 | External source (constraint solver) |

### The Fix

**Don't generate head nogoods for external atoms.**

| Atom Type | Generate δ_P(a)? | Generate Δ_P(a)? |
|-----------|------------------|------------------|
| Regular (a ∈ 𝒜) | ✓ Yes | ✓ Yes |
| External (a ∈ 𝒞) | ✗ No | ✗ No |

---

## Modified Completion: Δ^𝒞_P

The completion nogoods with externals:
```
Δ^𝒞_P = {body nogoods for ALL bodies}
